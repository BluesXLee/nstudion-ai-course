```python
def dot_product(a, b):
    if len(a) != len(b):
        raise ValueError("兩個向量的長度必須相同")

    total = 0
    for i in range(len(a)):
        total += a[i] * b[i]
    return total
```

```python
def matrix_shape(matrix):
    return (len(matrix), len(matrix[0]))
```

```python
def matmul(a, b):
    rows_a, cols_a = matrix_shape(a)
    rows_b, cols_b = matrix_shape(b)
    print(matrix_shape(A), matrix_shape(B))

    if cols_a != rows_b:
        raise ValueError("矩陣的內側維度必須相同")

    result = []
    for i in range(rows_a):
        row = []
        for j in range(cols_b):
            column_b = [b[k][j] for k in range(rows_b)]
            row.append(dot_product(a[i], column_b))
        result.append(row)

    return result
```
